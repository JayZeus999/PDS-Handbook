Boolean Indexing

```python
names = np.array(['Bob', 'Joe', 'Will', 'Bob', 'Will', 'Joe', 'Joe'])
```

```python
data = np.random.randn(7, 4)
```

```python
names
```

```python
data
```

```python
# Like arithmetic operations, comparisons (such as ==) with arrays are also vectorized. 
names == "Bob"
```

```python
# The boolean array must be of the same length as the array axis it’s indexing.
data[names == "Bob"]
```

```python
data[names == "Bob", 2:]
```

```python
# Only 1 column is selected, so both entries are put in the same list.
data[names == "Bob", 3]
```

```python
data[names == "Bob", 3:]
```

```python
names != "Bob"
```

```python
data[~(names == "Bob")]
```

```python
# Invert a general condition.
cond = names == "Bob"
```

```python
data[~cond]
```

```python
# To combine multiple boolean conditions, use & or |

mask = (names == "Bob") | (names == "Will")
```

```python
mask
```

```python
data[mask]
```

Selecting data from an array by boolean indexing always creates a copy of
the data, even if the returned array is unchanged.

```python
# To set all of the negative values in data to 0:
data[data < 0] = 0
```

```python
data
```

```python
data[names != "Joe"] = 7

data
```

```python
arr = np.arange(15).reshape((3, 5))

arr
```

```python
arr.T # (3 x 6)
```

```python
# Computing the inner matrix product: (Am×n​) ⋅ (Bn×p​)
# The inside dimensions must match; result shape becomes: (m×p)

np.dot(arr.T, arr) 
```

For higher dimensional arrays, transpose will accept a tuple of axis
numbers to permute the axes.

```python
arr = np.arange(16).reshape((2, 2, 4))

arr
```

Page 0:
row 0 -> [0 1 2 3]
row 1 -> [4 5 6 7]

Page 1:
row 0 -> [ 8  9 10 11]
row 1 -> [12 13 14 15]

Swap axis 0 and axis 1 --> Take axis 1 first, axis 0 second, axis 2 third (stays unchanged).

In simpler terms: Rows become pages, and pages become rows.

New Page 0: Take all the old row 0s.

New Page 1: Take all the old row 1s.

```python
arr.transpose((1, 0, 2))
```

```python
arr
```

```python
arr.swapaxes(1, 2)
```

A universal function, or ufunc, is a function that performs element-wise
operations on data in ndarrays.

```python
arr = np.arange(10)

arr
```

```python
# Unary ufuncs
np.sqrt(arr)
```

```python
# Take the exponential e^x
np.exp(arr)
```

```python
# Binary ufuncs

x = np.random.randn(8)

y = np.random.randn(8)
```

```python
x
```

```python
y
```

```python
np.maximum(x, y)
```

While not common, a ufunc can return multiple arrays.

```python
arr = np.random.randn(7) * 5

arr
```

```python
remainder, whole_part = np.modf(arr)
```

```python
remainder
```

```python
whole_part
```

Ufuncs accept an optional out argument that allows them to operate in
place on arrays:

```python
arr
```

```python
np.sqrt(arr)
```

```python
# The second arr is the out argument. --> “Store the result back inside arr itself.”
np.sqrt(arr, arr)
```
